# Week 3 — Regularization

Regularization discourages the model from using excessively large weights.

For L2 regularization, the cost adds a penalty proportional to:

$$\sum_j w_j^2$$

The regularization strength is controlled by `lambda`.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
w = np.array([0.5, -2.0, 4.0, 0.2])

for lambda_ in [0.0, 0.1, 1.0, 10.0]:
    penalty = (lambda_ / 2) * np.sum(w**2)
    print(f'lambda={lambda_:>4}: penalty={penalty:.4f}')
```

## Effect of lambda

- `lambda = 0` → no regularization
- small positive lambda → mild penalty
- very large lambda → strong pressure toward small weights

Too much regularization can cause underfitting.


```python
lambdas = np.linspace(0, 5, 100)
penalties = [(l/2)*np.sum(w**2) for l in lambdas]
plt.plot(lambdas, penalties)
plt.xlabel('lambda')
plt.ylabel('Regularization penalty')
plt.title('Effect of Regularization Strength')
plt.show()
```
