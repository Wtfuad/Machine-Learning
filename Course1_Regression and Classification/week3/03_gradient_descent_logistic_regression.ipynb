# Week 3 — Gradient Descent for Logistic Regression

Gradient descent updates `w` and `b` just like in linear regression, but predictions come from the sigmoid function.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def predict_probability(X, w, b):
    return sigmoid(X @ w + b)
```

```python
def compute_cost(X, y, w, b):
    predictions = predict_probability(X, w, b)
    predictions = np.clip(predictions, 1e-15, 1 - 1e-15)
    return -np.mean(y*np.log(predictions) + (1-y)*np.log(1-predictions))
```

```python
def compute_gradient(X, y, w, b):
    m = len(y)
    errors = predict_probability(X, w, b) - y
    dj_dw = (X.T @ errors) / m
    dj_db = np.sum(errors) / m
    return dj_dw, dj_db
```

```python
def gradient_descent(X, y, w, b, alpha=0.1, iterations=5000):
    history = []
    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        w = w - alpha * dj_dw
        b = b - alpha * dj_db
        cost = compute_cost(X, y, w, b)
        history.append(cost)
        if i % 1000 == 0:
            print(f'Iteration {i:5d}: cost={cost:.6f}')
    return w, b, history
```

```python
X = np.array([[0.5], [1.0], [1.5], [2.0], [3.0], [3.5], [4.0], [5.0]])
y = np.array([0, 0, 0, 0, 1, 1, 1, 1])

w0 = np.zeros(X.shape[1])
b0 = 0.0

w, b, history = gradient_descent(X, y, w0, b0)
print('w:', w)
print('b:', b)
```

```python
plt.plot(history)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.title('Logistic Regression Training')
plt.show()
```

```python
def predict_class(X, w, b, threshold=0.5):
    return (predict_probability(X, w, b) >= threshold).astype(int)

predictions = predict_class(X, w, b)
accuracy = np.mean(predictions == y)
print('Predictions:', predictions)
print('Accuracy:', accuracy)
```
