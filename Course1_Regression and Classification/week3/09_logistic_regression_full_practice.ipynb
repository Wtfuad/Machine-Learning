# Week 3 — Full Logistic Regression Practice

This is an original end-to-end practice notebook designed to prepare for the Week 3 logistic regression programming assignment without reproducing its graded solution.

Workflow:
1. create a binary classification dataset
2. visualize it
3. normalize features
4. train logistic regression from scratch
5. evaluate predictions
6. compare with scikit-learn


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix
```

```python
np.random.seed(42)
class0 = np.random.normal(loc=[2,2], scale=[0.7,0.7], size=(60,2))
class1 = np.random.normal(loc=[5,5], scale=[0.8,0.8], size=(60,2))
X = np.vstack([class0, class1])
y = np.hstack([np.zeros(60), np.ones(60)])
```

```python
plt.scatter(X[y==0,0], X[y==0,1], label='Class 0')
plt.scatter(X[y==1,0], X[y==1,1], label='Class 1')
plt.legend()
plt.show()
```

```python
mu = X.mean(axis=0)
sigma = X.std(axis=0)
X_norm = (X - mu) / sigma
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))

def cost(X, y, w, b):
    p = np.clip(sigmoid(X@w+b), 1e-15, 1-1e-15)
    return -np.mean(y*np.log(p)+(1-y)*np.log(1-p))

def gradient(X, y, w, b):
    m = len(y)
    e = sigmoid(X@w+b) - y
    return (X.T@e)/m, np.sum(e)/m

def train(X, y, alpha=0.1, iterations=5000):
    w = np.zeros(X.shape[1])
    b = 0.0
    history = []
    for _ in range(iterations):
        dw, db = gradient(X, y, w, b)
        w -= alpha*dw
        b -= alpha*db
        history.append(cost(X, y, w, b))
    return w, b, history
```

```python
w, b, history = train(X_norm, y)
pred_prob = sigmoid(X_norm@w+b)
pred = (pred_prob >= 0.5).astype(int)
print('From-scratch accuracy:', np.mean(pred == y))
print('w:', w, 'b:', b)
```

```python
plt.plot(history)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.show()
```

```python
sk_model = LogisticRegression()
sk_model.fit(X_norm, y)
sk_pred = sk_model.predict(X_norm)
print('Scikit-learn accuracy:', accuracy_score(y, sk_pred))
print('Confusion matrix:\n', confusion_matrix(y, sk_pred))
```

## Week 3 Takeaways
- sigmoid converts linear scores into probabilities
- logistic loss measures classification error
- gradient descent optimizes logistic regression parameters
- decision boundaries separate predicted classes
- overfitting harms generalization
- regularization discourages overly large weights
- regularized linear and logistic regression use the same core idea
