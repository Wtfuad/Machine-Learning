# Week 3 — Decision Boundary

The decision boundary separates regions where the model predicts class 0 and class 1.

For logistic regression with threshold 0.5:

$$w\cdot x + b = 0$$

defines the decision boundary.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([
    [1.0, 1.0], [1.5, 2.0], [2.0, 1.5], [2.0, 2.5],
    [4.0, 4.0], [4.5, 5.0], [5.0, 4.0], [5.5, 5.5]
])
y = np.array([0,0,0,0,1,1,1,1])
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_gradient(X, y, w, b):
    m = len(y)
    errors = sigmoid(X @ w + b) - y
    return (X.T @ errors)/m, np.sum(errors)/m

def gradient_descent(X, y, alpha=0.1, iterations=5000):
    w = np.zeros(X.shape[1])
    b = 0.0
    for _ in range(iterations):
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        w -= alpha * dj_dw
        b -= alpha * dj_db
    return w, b

w, b = gradient_descent(X, y)
print('w:', w, 'b:', b)
```

```python
plt.scatter(X[y==0,0], X[y==0,1], label='Class 0')
plt.scatter(X[y==1,0], X[y==1,1], label='Class 1')

x1_vals = np.linspace(0.5, 6, 100)
x2_vals = -(w[0]*x1_vals + b) / w[1]
plt.plot(x1_vals, x2_vals, label='Decision boundary')

plt.xlabel('x1')
plt.ylabel('x2')
plt.legend()
plt.show()
```
