# Week 3 — The Problem of Overfitting

A model underfits when it is too simple to capture the pattern.

A model overfits when it fits the training data extremely closely but does not generalize well to new data.

We will visualize the idea using polynomial models.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
np.random.seed(4)
x = np.linspace(-3, 3, 30)
y = 2*x**2 + x + 3 + np.random.normal(0, 2.5, len(x))

indices = np.arange(len(x))
np.random.shuffle(indices)
train_idx = indices[:20]
test_idx = indices[20:]

x_train, y_train = x[train_idx], y[train_idx]
x_test, y_test = x[test_idx], y[test_idx]
```

```python
degrees = [1, 2, 12]

for degree in degrees:
    poly = PolynomialFeatures(degree=degree, include_bias=False)
    X_train_poly = poly.fit_transform(x_train.reshape(-1,1))
    X_test_poly = poly.transform(x_test.reshape(-1,1))

    model = LinearRegression()
    model.fit(X_train_poly, y_train)

    train_mse = mean_squared_error(y_train, model.predict(X_train_poly))
    test_mse = mean_squared_error(y_test, model.predict(X_test_poly))
    print(f'Degree {degree:2d}: train MSE={train_mse:.3f}, test MSE={test_mse:.3f}')
```

## Interpretation

- Degree 1 may underfit.
- A moderate degree can capture the true pattern.
- A very high degree may achieve a tiny training error while producing a worse test error.

That gap is a sign of poor generalization.
