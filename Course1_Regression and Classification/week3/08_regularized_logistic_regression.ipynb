# Week 3 — Regularized Logistic Regression

Regularized logistic regression adds an L2 penalty to the cost.

The bias `b` is normally not regularized.


```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def compute_cost_regularized(X, y, w, b, lambda_=1.0):
    m = len(y)
    p = sigmoid(X @ w + b)
    p = np.clip(p, 1e-15, 1-1e-15)
    base_cost = -np.mean(y*np.log(p) + (1-y)*np.log(1-p))
    reg_cost = (lambda_/(2*m)) * np.sum(w**2)
    return base_cost + reg_cost

def compute_gradient_regularized(X, y, w, b, lambda_=1.0):
    m = len(y)
    errors = sigmoid(X @ w + b) - y
    dj_dw = (X.T @ errors)/m + (lambda_/m)*w
    dj_db = np.sum(errors)/m
    return dj_dw, dj_db
```

```python
def gradient_descent_regularized(X, y, lambda_=1.0, alpha=0.1, iterations=5000):
    w = np.zeros(X.shape[1])
    b = 0.0
    history = []

    for _ in range(iterations):
        dj_dw, dj_db = compute_gradient_regularized(X, y, w, b, lambda_)
        w -= alpha*dj_dw
        b -= alpha*dj_db
        history.append(compute_cost_regularized(X, y, w, b, lambda_))

    return w, b, history
```

```python
X = np.array([[1,1],[1.5,2],[2,1.5],[2,2.5],[4,4],[4.5,5],[5,4],[5.5,5.5]], dtype=float)
y = np.array([0,0,0,0,1,1,1,1])

for lambda_ in [0.0, 0.1, 1.0, 10.0]:
    w, b, history = gradient_descent_regularized(X, y, lambda_=lambda_)
    print(f'lambda={lambda_:>4}: w={w}, b={b:.4f}, final cost={history[-1]:.4f}')
```
