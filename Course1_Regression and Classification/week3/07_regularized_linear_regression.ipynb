# Week 3 — Regularized Linear Regression

We compare ordinary polynomial regression with Ridge regression, which applies L2 regularization.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Ridge
```

```python
np.random.seed(7)
x = np.linspace(-3, 3, 25)
y = 2*x**2 + x + 3 + np.random.normal(0, 3, len(x))

poly = PolynomialFeatures(degree=10, include_bias=False)
X_poly = poly.fit_transform(x.reshape(-1,1))
```

```python
ordinary = LinearRegression()
ordinary.fit(X_poly, y)

ridge = Ridge(alpha=10.0)
ridge.fit(X_poly, y)

print('Ordinary weight norm:', np.linalg.norm(ordinary.coef_))
print('Ridge weight norm:', np.linalg.norm(ridge.coef_))
```

```python
x_plot = np.linspace(-3, 3, 300)
X_plot = poly.transform(x_plot.reshape(-1,1))

plt.scatter(x, y, label='Data')
plt.plot(x_plot, ordinary.predict(X_plot), label='No regularization')
plt.plot(x_plot, ridge.predict(X_plot), label='Ridge regularization')
plt.legend()
plt.show()
```
