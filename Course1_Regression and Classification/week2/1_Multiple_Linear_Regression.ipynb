Multiple Linear Regression

In Week 1, the model had one feature:

f(x) = wx + b

Now we use multiple features:

f(x) = w1x1 + w2x2 + ... + wnxn + b

Using vectors:

f(x) = w · x + b

```python
import numpy as np
import matplotlib.pyplot as plt
```

Example Dataset

We will predict house prices using:

- size
- bedrooms
- floors
- age

```python
X_train = np.array([
    [2104, 5, 1, 45],
    [1416, 3, 2, 40],
    [1534, 3, 2, 30],
    [852, 2, 1, 36],
    [1940, 4, 2, 20]
], dtype=float)

y_train = np.array([
    460,
    232,
    315,
    178,
    410
], dtype=float)

print("X shape:", X_train.shape)
print("y shape:", y_train.shape)
```

Prediction Function

```python
def predict_single(x, w, b):
    return np.dot(x, w) + b
```

```python
w = np.array([0.1, 10, -20, -1])
b = 50
prediction = predict_single(X_train[0], w, b)
print("Prediction =", prediction)
print("Actual =", y_train[0])
```

Vectorized Predictions for All Examples

```python
def predict(X, w, b):
    return X @ w + b
```

```python
predictions = predict(X_train, w, b)
print(predictions)
```

Cost Function

```python
def compute_cost(X, y, w, b):
    m = len(y)
    predictions = X @ w + b
    errors = predictions - y
    cost = np.sum(errors ** 2) / (2 * m)
    return cost

cost = compute_cost(X_train, y_train, w, b)
print("Cost =", cost)
```

Key Takeaways

- Multiple linear regression uses multiple input features.
- X is a matrix.
- w is a vector.
- One prediction is w · x + b.
- Predictions for all examples can be computed using X @ w + b.
