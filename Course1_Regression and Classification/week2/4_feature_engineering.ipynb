Week 2 — Feature Engineering

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
length = np.array([20,25,30,35,40], dtype=float)
width = np.array([15,18,20,22,25], dtype=float)
area = length * width
X = np.column_stack([length, width, area])
print(X)
```

```python
distance = np.array([2,4,6,8,10], dtype=float)
traffic = np.array([1.0,1.2,1.5,1.8,2.0])
speed = np.array([25,24,22,20,18], dtype=float)
distance_traffic = distance * traffic
estimated_travel_time = distance / speed
X_engineered = np.column_stack([distance, traffic, speed, distance_traffic, estimated_travel_time])
print(X_engineered)
```
