Week 2 — Feature Scaling

Use z-score normalization to place features on similar scales.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train = np.array([[2104,5,45],[1416,3,40],[1534,3,30],[852,2,36],[1940,4,20]], dtype=float)
```

```python
def zscore_normalize(X):
    mu = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    return (X - mu) / sigma, mu, sigma

X_norm, mu, sigma = zscore_normalize(X_train)
print('mu:', mu)
print('sigma:', sigma)
print('normalized:\n', X_norm)
```

```python
new_house = np.array([1600,3,25], dtype=float)
print('normalized new house:', (new_house - mu) / sigma)
```
