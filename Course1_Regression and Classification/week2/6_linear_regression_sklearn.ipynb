Week 2 — Linear Regression with Scikit-learn

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
X_train = np.array([[2104,5,1,45],[1416,3,2,40],[1534,3,2,30],[852,2,1,36],[1940,4,2,20],[1200,3,1,15],[1700,4,2,10],[1000,2,1,50]], dtype=float)
y_train = np.array([460,232,315,178,410,250,390,190], dtype=float)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_train)
print('weights:', model.coef_)
print('bias:', model.intercept_)
print('mse:', mean_squared_error(y_train, predictions))
```

```python
new_house = np.array([[1600,3,2,20]], dtype=float)
print('predicted price:', model.predict(new_house)[0])
```

```python
manual_predictions = X_train @ model.coef_ + model.intercept_
print('matches sklearn:', np.allclose(predictions, manual_predictions))
```
