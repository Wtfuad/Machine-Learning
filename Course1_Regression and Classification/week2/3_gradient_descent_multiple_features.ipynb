Week 2 — Gradient Descent with Multiple Features

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X_train = np.array([[2104,5,1,45],[1416,3,2,40],[1534,3,2,30],[852,2,1,36],[1940,4,2,20],[1200,3,1,15],[1700,4,2,10],[1000,2,1,50]], dtype=float)
y_train = np.array([460,232,315,178,410,250,390,190], dtype=float)
mu = np.mean(X_train, axis=0)
sigma = np.std(X_train, axis=0)
X_norm = (X_train - mu) / sigma
```

```python
def predict(X, w, b):
    return X @ w + b

def compute_cost(X, y, w, b):
    m = len(y)
    errors = predict(X, w, b) - y
    return np.sum(errors**2)/(2*m)

def compute_gradient(X, y, w, b):
    m = len(y)
    errors = predict(X, w, b) - y
    return (X.T @ errors)/m, np.sum(errors)/m
```

```python
def gradient_descent(X, y, w, b, alpha, iterations):
    history = []
    for i in range(iterations):
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        w = w - alpha*dj_dw
        b = b - alpha*dj_db
        history.append(compute_cost(X, y, w, b))
    return w, b, history
```

```python
w0 = np.zeros(X_norm.shape[1])
b0 = 0.0
w, b, history = gradient_descent(X_norm, y_train, w0, b0, 0.01, 10000)
print('w:', w)
print('b:', b)
plt.plot(history)
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.show()
```
